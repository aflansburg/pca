# VBC: County Aggregate Expenditure & Risk Score — Delaware peek

Scope: county-level Medicare-FFS expenditure + HCC risk on assignable beneficiaries (Q3 backdrop; HISTORY 002).
Source: `data/sources/vbc/County_level_Aggregate_Expenditure_and_Risk_Score_Data_on_Assignable_Beneficiaries_2025_oc1.zip`.
Reading: small file, read whole, filter `STATE_NAME` = DELAWARE.

```python
import zipfile
from pathlib import Path

import pandas as pd

ROOT = next(p for p in [Path.cwd(), *Path.cwd().parents] if (p / "pyproject.toml").exists())
ZIP = ROOT / "data/sources/vbc/County_level_Aggregate_Expenditure_and_Risk_Score_Data_on_Assignable_Beneficiaries_2025_oc1.zip"
MEMBER = "County_level_Aggregate_Expenditure_and_Risk_Score_Data_on_Assignable_Beneficiaries_2025_oc1.csv"
STATE_COL, STATE = "STATE_NAME", "DELAWARE"
with zipfile.ZipFile(ZIP) as z, z.open(MEMBER) as f:
    df = pd.read_csv(f, dtype=str, encoding="latin-1")
de = df[df[STATE_COL] == STATE].copy()
print("all rows:", df.shape, "| DE rows:", de.shape)
de[["COUNTY_NAME", "PER_CAPITA_EXP_AGND", "AVG_RISK_SCORE_AGND", "PERSON_YEARS_AGND"]]
```

## Notes
- Grain: one row per county (Kent, New Castle, Sussex). Enrollment-type suffixes: ESRD, DIS, AGDU, AGND.
- Medicare-FFS assignable population only (no MA/Medicaid/commercial). A backdrop, not an all-payer denominator (HISTORY 002).
- `AGND` = aged non-dual (largest group). Use `PER_CAPITA_EXP_*` + `AVG_RISK_SCORE_*` for the Q3 population-adjusted context.
