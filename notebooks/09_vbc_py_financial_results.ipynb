```python
import zipfile
from pathlib import Path

import pandas as pd

ROOT = next(p for p in [Path.cwd(), *Path.cwd().parents] if (p / "pyproject.toml").exists())
ZIP = ROOT / "data/sources/vbc/PY_Financial_and_Quality_Results_2024.zip"
MEMBER = "PY_Financial_and_Quality_Results_2024.csv"
with zipfile.ZipFile(ZIP) as z, z.open(MEMBER) as f:
    df = pd.read_csv(f, dtype=str, encoding="latin-1")
print("shape:", df.shape, "| unique ACOs:", df["ACO_ID"].nunique())
print("has county/state column:", any(c.lower() in ("state", "county") for c in df.columns))
df[["ACO_ID", "ACO_Name", "N_AB", "QualScore", "GenSaveLoss", "EarnSaveLoss"]].head(10)
```
